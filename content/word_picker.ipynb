Notebook used to prototype a word picker from the list of the 1000 most common words. 

```python
import pandas as pd
import random
```

```python
word_list = pd.read_csv("italian_words_frequency.csv")
word_list.head()
```

```python
word_list['is title'] = word_list['word'].apply(lambda x: x[0].isupper())
word_list['is title'].value_counts()
```

```python
word_list = word_list[word_list['is title'] == False].reset_index(drop=True)
word_list['cumulative frequency'] = word_list['frequency'].cumsum()
word_list
```

```python

word_list
```

```python


def pick_word(word_list):
    total = word_list['cumulative frequency'].iloc[-1]
    r = random.randint(1, total)
    return word_list[word_list['cumulative frequency'] >= r].iloc[0]['word']
print(pick_word(word_list))
```

```python
for i in range(10):
    print(pick_word(word_list))
```

```python
word_list[['rank', 'word', 'frequency', 'cumulative frequency']].to_csv('word_list.csv', index=False)
```
